#### Pretraining

```python
from poetron import Poetron

p = Poetron()
p.pretrain(batch_size=64, epochs=10, log_epochs=1)
```

```python
# delete Poetron instance
del p
```

#### Generated Poems

```python
# create new Poetron instance, load saved model and tokenizer
p = Poetron()
p.load()
```

```python
# starting with empty poems
poems = p.generate(['', '', ''], max_new_tokens=180, postprocess=True, top_k=6)
```

```python
print(poems[0])
```

```python
print(poems[1])
```

```python
print(poems[2])
```

```python
# starting with partially written poems
poems = p.generate([
    '<poem_start>on a nice, warm day',
    '<poem_start>over the mountains',
    '<poem_start>snowfall beside me'],
    max_new_tokens=180, postprocess=True, top_k=6)
```
